```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn import datasets     # iris 데이터 세트
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = datasets.load_iris()
data
# iris 데이터세트 읽기
# data 부분에 있는 feature_names을 열 이름으로 하는 데이터 프레임
df = pd.DataFrame(data['data'], columns=data['feature_names'])
df['target'] = data['target'] # target 부분의 데이터를 데이터 프레임에 열로 추가
df.head()
# 데이터 프레임 앞부분 출력
plt.figure(figsize=(8, 8)) # 그래프 출력
ax = sns.pairplot(df, hue='target')
plt.show()
X_df = df.iloc[:, [0,1,2,3]] # data 부분
print(X_df.mean(axis=0)) # 열별 평균
print(X_df.var(axis=0)) # 열별 분산
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.metrics import confusion_matrix
# 혼동 행렬
from sklearn.metrics import accuracy_score # 분류 정확도
iris = datasets.load_iris()
print(iris.feature_names)
print(iris.target_names)
data = iris.data
target = iris.target
print(data)
print(target)
plt.figure(figsize=(8, 4))
plt.boxplot([data[:,0], data[:,1], data[:,2], data[:,3]], labels=iris.feature_names)
plt.title('Iris')
plt.xlabel('Features')
plt.ylabel('Length or Width (cm)')
plt.show()
X = data[:, 0]
Y = data[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.show()
```

```python
scaler = StandardScaler()
data_std = scaler.fit_transform(data)
print(data_std)
plt.figure(figsize=(5, 5))
X = data_std[:, 0]
Y = data_std[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris (Standardization)')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.show()
TWSS = {}
for k in range(1, 11):
    model = KMeans(n_clusters=k)
    model.fit(data_std)
    TWSS[k] = model.inertia_
plt.plot(list(TWSS.keys()), list(TWSS.values()), marker='o')
plt.xlabel("Number of Clusters(K)")
plt.ylabel("TWSS")
plt.show()
```

```python
K=3
kmeans = KMeans(n_clusters=K)
kmeans.fit(data_std)
centers = kmeans.cluster_centers_
labels = kmeans.labels_
print(centers)
print(labels)
plt.figure(figsize=(5, 5))
X = data_std[:, 0]
Y = data_std[:, 1]
plt.scatter(X, Y, c=target)
plt.title('Iris (Standardization)')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
X = centers[:, 0]
Y = centers[:, 1]
plt.scatter(X, Y, c='red', s=200, alpha=0.5);
plt.show()
```

```python
print(target)
print(labels)
confusion_matrix(target, labels)
for i in range(len(labels)):
    if labels[i] == 1:
        labels[i] = 0
    elif labels[i] == 0:
        labels[i] = 1
# 군집화에 의해 분류된 라벨
print(labels)
# 교차표: 실제 종 대비 군집화 결과
print(confusion_matrix(target, labels))
# 정확하게 일치하는 개수
print(accuracy_score(target, labels, normalize=False))
# 정확도 (비율)
print(accuracy_score(target, labels))   # normalize=True
```

```python
from sklearn.datasets import load_iris
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
from matplotlib import pyplot as plt
iris = load_iris()
iris
print(iris.feature_names)
print(iris.target_names)
X = iris.data
y = iris.target
print(X.shape)
print(X)
print(y.shape)
print(y)
# 데이터 세트의 분리: 학습용 입력값, 테스트용 입력값, 학습용 목표값, 테스트용 목표값
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1234)
print(X_train.shape)
print(X_train)
print(y_train.shape)
print(y_train)
print(X_test.shape)
print(X_test)
print(y_test.shape)
print(y_test)
```

```python
# C5.0 모형 설정
clf = tree.DecisionTreeClassifier(criterion = "entropy")
# 학습
clf = clf.fit(X_train, y_train)
# 그래프 출력
plt.figure(figsize=(15, 12))
# 학습된 모형 출력
tree.plot_tree(clf)
plt.show()
# 분류 예측
y_pred = clf.predict(X_test)
print(y_pred)
# 평가: 혼동 행렬 출력
print(confusion_matrix(y_test, y_pred))
# 정확도 출력
print("정확도:", accuracy_score(y_test, y_pred))
```
